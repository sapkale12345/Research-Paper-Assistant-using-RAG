```python
pdf_path = "/content/drive/MyDrive/Colab Notebooks/Research Paper/Attention Is All You Need.pdf"

loader = PyMuPDFLoader(pdf_path)

documents = loader.load()
print("Total Pages:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
```

```python
chunks = text_splitter.split_documents(documents)
print("Total Chunks:", len(chunks))
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_chroma import Chroma
vector_db = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory="/content/drive/MyDrive/Colab Notebooks/Research Paper/chroma_db"
)
```

```python
print("Total Chunks:", vector_db._collection.count())
```

```python
retriever = vector_db.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 3,
        "fetch_k": 10
    }
)
```

```python
question = "What is self-attention?"

docs = retriever.invoke(question)
```

```python
for i, doc in enumerate(docs, start=1):

    print("=" * 60)

    print(f"Chunk {i}")

    print()

    print(doc.page_content)

    print()

    print(doc.metadata)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages(

    [

        (
            "system",

            """
You are an AI Research Paper Assistant.

Your job is to answer questions ONLY from the retrieved context.

Instructions:

1. Answer ONLY from the retrieved context.
2. Do not use your own knowledge.
3. If the answer is not available, reply:

"I could not find this information in the research paper."

4. Keep answers clear and concise.
5. Explain technical concepts in simple language whenever possible.
6.At the end of the answer, do NOT mention page numbers. They will be displayed separately.

Retrieved Context:
{context}

"""
        ),

        (
            "human",

            "{question}"
        )

    ]

)
```

```python
chain = prompt | llm
```

```python
question = input("Ask your question: ")
```

```python
docs = retriever.invoke(question)

context = "\n\n".join(
    doc.page_content for doc in docs
)
```

```python
response = chain.invoke(
    {
        "context": context,
        "question": question
    }
)
```

```python
print("\nAnswer:\n")
print(response.content)
pages = sorted(
    set(doc.metadata["page"] + 1 for doc in docs)
)

print("\nSource Pages:", pages)
```

```python
def answer_question(question):

    docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    response = chain.invoke(
        {
            "context": context,
            "question": question
        }
    )

    pages = sorted(
        set(doc.metadata["page"] + 1 for doc in docs)
    )

    return response.content, pages
```

```python
while True:

    question = input("\nAsk your question (type 'exit' to quit): ")

    if question.lower() == "exit":
        break

    answer, pages = answer_question(question)

    print("\nAnswer:\n")
    print(answer)

    print("\nSource Page(s):", pages)
```
